# ⚔️ Global Arsenal Intelligence

## Visão Geral

Análise exploratória de mais de 10.000 sistemas de armas militares distribuídos em 128 países,
cobrindo um período histórico de 90 anos (1933–2024). Este projeto aplica técnicas de Data Analytics
para extrair padrões estratégicos, tecnológicos e geopolíticos a partir de dados de defesa publicamente disponíveis.

---

## 🎯 Objetivos

### 1. Análise Exploratória dos Dados (EDA)
Compreender a estrutura do dataset, identificar distribuições, outliers e padrões gerais
nos sistemas de armas registrados.

### 2. Evolução Tecnológica ao Longo do Tempo
Investigar como as especificações técnicas dos sistemas — como velocidade, alcance e custo —
evoluíram entre 1933 e 2024.

### 3. Comparação de Capacidades por País e Aliança
Analisar e comparar o arsenal militar de diferentes nações, com foco em países membros da OTAN
versus não-membros.

### 4. Análise de Exportação e Transferência de Armas
Identificar os principais exportadores globais de sistemas militares e os padrões
de adoção entre países operadores.

### 5. Segmentação e Clustering de Sistemas
Agrupar sistemas de armas por perfil técnico e operacional utilizando algoritmos
de aprendizado de máquina não supervisionado.

### 6. Análise de Texto — Coluna `Notes`
Aplicar técnicas de NLP para extrair informações relevantes das descrições textuais
de cada sistema de arma.

### 7. Análise de Custo e Valor Estratégico
Relacionar o custo unitário (`Unit_Cost_USD`) com desempenho, adoção internacional
e histórico de combate comprovado.
